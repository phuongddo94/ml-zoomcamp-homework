```python
asia_car_df = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].copy()
```

```python
X = asia_car_df.head(7).values
X
```

```python
# compute dot product of XTX
XTX = X.T.dot(X)
XTX
```

```python
# compute the inverse matrix of XTX
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100,1200])
```

```python
w = (XTX_inv @ X.T) @ y
```
